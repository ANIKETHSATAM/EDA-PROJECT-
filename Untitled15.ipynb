```python

#outlier analsyis
```

```python
rows=[]
for c in df.select_dtypes(np.number):
    q1,q3=df[c].quantile([.25,.75]); iqr=q3-q1;lo,hi=q1-1.5*iqr,q3+1.5*iqr
    n=((df[c]<lo)|(df[c]>hi)).sum(); rows.append([c,q1,q3,lo,hi,int(n),round(n/len(df)*100,2)])
    display(pd.DataFrame(rows,columns=["column","q1","q3","lower","upper","outlier_count","outlier_percent",]))
```

```python
#bivariate analysis
```

```python
for x,y in [("unit_price","quantity"),("unit_price","discount"),("quantity","discount")]:
    if x in df and y in df:
        plt.figure(figsize=(7,5));plt.scatter(df[x],df[y],aplha=.65);plt.xlabel(x);plt.ylabel(y);plt.title(f"{y} vs {x}"); plt.show()
for c in ["category","city","product_name"]:
    if c in df and "profit" in df: display(df.groupby(c).profit.agg(["sum","mean","count"]).sort_values("sum",ascending=False))
```

```python
import matplotlib.pyplot as plt

pairs = [
    ("unit_price", "quantity"),
    ("unit_price", "discount"),
    ("quantity", "discount")
]

for x, y in pairs:
    if x in df.columns and y in df.columns:
        plt.figure(figsize=(7,5))
        plt.scatter(df[x], df[y], alpha=0.65)
        plt.xlabel(x)
        plt.ylabel(y)
        plt.title(f"{y} vs {x}")
        plt.show()
```

```python
if "quantity" in df.columns and "discount" in df.columns:
    summary = (
        df.groupby("quantity")["discount"]
          .agg(["sum", "mean", "count"])
          .sort_values("sum", ascending=False)
    )
    display(summary)
```

```python
if "product" in df.columns and "discount" in df.columns:
    summary = (
        df.groupby("product")["quantity"]
          .agg(["sum", "mean", "count"])
          .sort_values("sum", ascending=False)
    )
    display(summary)
```

```python
if "quantity" in df.columns and "discount" in df.columns:
    summary = (
        df.groupby("product")["discount"]
          .agg(["sum", "mean", "count"])
          .sort_values("sum", ascending=False)
    )
    display(summary)
```
